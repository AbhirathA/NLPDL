**Loading Glove**

```python
glove_file = 'glove.6B.50d.txt'
embeddings_index = {}
with open(glove_file, 'r', encoding='utf-8') as f:
    for line in f:
        values = line.split()
        word = values[0]
        coefs = numpy.asarray(values[1:], dtype='float32')
        embeddings_index[word] = coefs

print(f'Found {len(embeddings_index)} word vectors in {glove_file}')
```

**Q2) Part B**

```python
def cosine_similarity(embedding1, embedding2):
  w1 = numpy.linalg.norm(embedding1)
  w2 = numpy.linalg.norm(embedding2)
  if(w1 == 0 or w2 == 0):
    return 0
  return numpy.dot(embedding1, embedding2) / (w1 * w2)
```

```python
father = embeddings_index['father']
mother = embeddings_index['mother']

ball = embeddings_index['ball']
crocodile = embeddings_index['crocodile']

italy = embeddings_index['italy']
rome = embeddings_index['rome']
france = embeddings_index['france']
paris = embeddings_index['paris']
```

```python
print(f'Cosine Similarity between father and mother: {cosine_similarity(father, mother)}')
print(f'Cosine Similarity between ball and crocodile: {cosine_similarity(ball, crocodile)}')
print(f'Cosine Similarity between france-paris and rome-italy: {cosine_similarity(france-paris, rome-italy)}')
```

**Q2) Part C**

```python
def complete_analogy_method1(word1, word2, word3):
  embedding1 = embeddings_index[word1]
  embedding2 = embeddings_index[word2]
  embedding3 = embeddings_index[word3]

  vec1 = embedding2-embedding1
  ans = word3
  ansVal = cosine_similarity(vec1, embedding3-embedding3)
  for (i,v) in embeddings_index.items():
    t = cosine_similarity(vec1, v - embedding3)
    if(t >= ansVal):
      ans = i
      ansVal = t
  return ans
```

```python
def complete_analogy_method2(word1, word2, word3):
  embedding1 = embeddings_index[word1]
  embedding2 = embeddings_index[word2]
  embedding3 = embeddings_index[word3]
  vec1 = embedding2 - embedding1 + embedding3
  ans = word3
  ansVal = cosine_similarity(vec1, embedding3)
  for (i,v) in embeddings_index.items():
    t = cosine_similarity(vec1, v)
    if(t >= ansVal):
      ans = i
      ansVal = t
  return ans
```

```python
triads_to_try = [('italy', 'italian', 'spain'), ('india', 'delhi', 'japan'), ('man', 'woman', 'boy'), ('small', 'smaller', 'large')]
```

```python
print("Method 1:")
for triad in triads_to_try:
  print ('{} -> {} :: {} > {}'.format(*triad, complete_analogy_method1(*triad)))
```

```python
print("Method 2:")
for triad in triads_to_try:
  print ('{} -> {} :: {} > {}'.format(*triad, complete_analogy_method2(*triad)))
```

**Q2) Part E**

```python
def project(vec, axis):
  a = numpy.linalg.norm(axis)
  if(a == 0):
    return numpy.zeros(len(vec))
  return (numpy.dot(vec, axis) / numpy.linalg.norm(axis)) * (axis/numpy.linalg.norm(axis))
```

```python
def remove_bias(vec, axis):
  return vec - project(vec,axis)
```

```python
words_to_debias = ["receptionist"]
bias_axis = gender_axis
```

```python
for i in words_to_debias:
  print(f'Word: {i}.\nSimilarity along gender_axis: {cosine_similarity(embeddings_index[i], bias_axis)}')
  print("De-biasing now....")
  print(f"Similarity after De-biasing: {cosine_similarity(remove_bias(embeddings_index[i],bias_axis), bias_axis)}")
```

**Q2) Part F**

```python
def equilize(embedding1, embedding2, axis):
  u = embedding1 + embedding2
  u = u/2
  u_parallel = project(u, axis)
  u_orthogonal = u - u_parallel

  e1_parallel = project(embedding1, axis)
  e2_parallel = project(embedding2, axis)

  e1_corrected = numpy.sqrt(numpy.abs(1-numpy.linalg.norm(u_orthogonal)**2))*(e1_parallel - u_parallel)/numpy.linalg.norm(embedding1 - u_parallel- u_orthogonal)
  e2_corrected = numpy.sqrt(numpy.abs(1-numpy.linalg.norm(u_orthogonal)**2))*(e2_parallel - u_parallel)/numpy.linalg.norm(embedding2 - u_parallel- u_orthogonal)

  return e1_corrected + u_orthogonal, e2_corrected + u_orthogonal
```

```python
e1 = embeddings_index['man']
e2 = embeddings_index['woman']

e1_corrected, e2_corrected = equilize(e1, e2, gender_axis)
r_norm = remove_bias(embeddings_index['receptionist'], gender_axis)

print("Comparing cosine similarities with axis:")

print("Man (Before):", cosine_similarity(e1, gender_axis))
print("Woman (Before):", cosine_similarity(e2, gender_axis))
print()
print("Man (After):", cosine_similarity(e1_corrected, gender_axis))
print("Woman (After):", cosine_similarity(e2_corrected, gender_axis))

print()

print("Comparing cosine similarities with receptionist (normalised):")

print("Man (Before):", cosine_similarity(e1, r_norm))
print("Woman (Before):", cosine_similarity(e2, r_norm))
print()
print("Man (After):", round(cosine_similarity(e1_corrected, r_norm),7))
print("Woman (After):", round(cosine_similarity(e2_corrected, r_norm),7))


```
